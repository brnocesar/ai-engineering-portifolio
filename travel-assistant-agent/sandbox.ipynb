## Conectando direto na API da OpenAI

```python
from openai import OpenAI
from dotenv import load_dotenv
import os
```

```python
# carrega chave da api para variaveis de ambiente do sistema
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")

model_version = "gpt-6-luna"
```

```python
# monta prompt para interacao com API da OpenAI
numero_dias     = 7
numero_criancas = 2
atividade       = "música"
prompt          = f"Crie um roteiro de viagem de {numero_dias} dias, para uma familia com {numero_criancas} crianças, que gosta de {atividade}"
```

```python
# define client que vai permitir interacao com API da OpenAI
client = OpenAI(api_key=api_key)
client
```

Agora que temos o _client_ definido, vamos testar algumar requisições para a [API](https://developers.openai.com/api/docs/guides/migrate-to-responses), primeiro com _chat completion_ e depois uma "resposta".

### _Chat Completion_

Esse formato é uma estrutura de interação muito específica da API da OpenAI. O _chat completion_ é o recurso projetado para receber mensagens de um histórico ou uma pergunta e retornar uma resposta gerada pelo modelo. É uma forma de comunicação que segue a lógica interna e as nomenclaturas da própria OpenAI.

```python
# cria chat completion para interacao
# parametro messages eh uma lista que precisa de ao menos um item, mas geralmente passamos dois
response = client.chat.completions.create(
    model=model_version,
    messages=[
        {"role": "system", "content": "Você é um assistente de roteiro de viagens."},   # um que indica como o sistema deve se comportar
        {"role": "user",   "content": prompt}                                           # outro com a pergunta do usuario
    ],
    store=False
)
type(response)
```

```python
for r in response:
    print(r)
```

```python
type(response.choices), len(response.choices)
```

```python
for choice in response.choices:
    for property in choice:
        print(property)
```

```python
type(response.choices[0].message)
```

```python
print(response.choices[0].message.content.replace("; ", ";\n"))
```

## LangChain

O [LangChain](https://docs.langchain.com/oss/python/langchain/overview) é um _framework_ focado em facilitar o desenvolvimento de aplicações que utilizam modelos de linguagem (LLMs). Ele funciona como uma camada de abstração que ajuda a orquestrar diferentes recursos, permitindo a construção de soluções mais estruturadas e profissionais.

### Usando o contrutor do LangChain

Vamos nos conectar a API da OpenAI através do LangChain, e dessa forma, será possível isolar a lógica da aplicação da implementação específica de um modelo. Ou seja, se um dia quiser trocar o modelo da OpenAI por outro disponível no mercado, a estrutura do código permanecerá praticamente a mesma. A flexibilidade e padronização do LangChain permitem que a aplicação não fique "presa" aos construtores ou métodos específicos de uma única empresa.

```python
from langchain_openai import ChatOpenAI
```

```python
# passa a usar construtor do langchain
model = ChatOpenAI(
    model=model_version,
    # temperature=0.5, # BadRequestError: 'temperature' does not support 0.5 with this model. Only the default (1) value is supported.
    api_key=api_key
)
```

```python
# agora usamos o metodo invoke, que recebe apenas o prompt
response = model.invoke(prompt)

type(response)
```

```python
for r in response:
    print(r)
```

```python
print(response.content)
```

A grande vantagem de usarmos o LangChain, como vimos nesta aula, é justamente a capacidade de abstrair esse tipo de particularidade. Ao utilizar o LangChain com o método invoke, nós não precisamos nos preocupar se estamos interagindo com um completion da OpenAI ou com outra interface proprietária de outro modelo. O LangChain padroniza essa comunicação.

Dessa forma, se você quiser trocar a OpenAI por outro fornecedor no futuro, o seu código continua funcionando com o invoke, sem que você precise reescrever a lógica para se adaptar às especificidades (como o Chat Completion) de cada empresa.

Faz sentido para você por que o LangChain atua como um "tradutor" que uniformiza essas chamadas, evitando que você precise aprender e implementar os detalhes técnicos exclusivos de cada API de mercado?

Ao usar o método `invoke` do LangChain, não precisamos nos preocupar se estamos interagindo com um _completion_ da OpenAI ou outra interface proprietária de algum modelo. O LangChain padroniza essa comunicação, permitindo isolar a lógica da aplicação da implementação específica de um modelo. Resumindo:

### Usando PromptTemplate

Dentro do contexto da LangChain, o [PromptTemplate](https://reference.langchain.com/python/langchain-core/prompts/prompt/PromptTemplate) é uma ferramenta utilizada para criar modelos estruturados de texto que serão enviados como instruções para o modelo de linguagem. Em vez de se escrever frases soltas ou utilizarmos f-_strings_ de forma repetida e manual a cada nova interação, o PromptTemplate permite definir um formato fixo com espaços reservados, chamados de variáveis.

O PromptTemplate funciona como um formulário pré-impresso: tem o texto principal que define as regras e o comportamento esperado da IA, mas deixa lacunas específicas para preencher com informações variáveis, como o número de dias de uma viagem ou a quantidade de crianças.

Essa abordagem traz vantagens importantes para o desenvolvimento:

- Organização: o código fica muito mais limpo, pois separa a estrutura da lógica de programação do conteúdo das instruções enviadas ao modelo.
- Consistência: ao padronizar o formato do _prompt_, se garante que a interação com a API do OpenAI siga sempre um padrão definido, facilitando a manutenção e a previsibilidade dos resultados.

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv
import os

load_dotenv()
api_key       = os.getenv("OPENAI_API_KEY")
model_version = "gpt-6-luna"
```

```python
# string que usamos anteriormnte como prompt
numero_dias     = 7
numero_criancas = 2
atividade       = "música"
prompt          = f"Crie um roteiro de viagem de {numero_dias} dias, para uma familia com {numero_criancas} crianças, que gosta de {atividade}"

print(prompt)
```

```python
# inicializa o prompt template definindo as variaveis necessarias
prompt_model = PromptTemplate(
    template=(
        "Crie um roteiro de viagem de {numero_dias} dias, "
        "para uma familia com {numero_criancas} crianças, "
        "que gosta de {atividade}"
    )
)
prompt_model
```

```python
# formata prompt passando as variaveis
prompt = prompt_model.format(
    numero_dias=numero_dias,
    numero_criancas=numero_criancas,
    atividade=atividade
)
print(prompt)
```

```python
model    = ChatOpenAI(model=model_version, api_key=api_key)
response = model.invoke(prompt)

print(response.content)
```

## LCEL e cadeias

Agora vamos começar a fazer uso do **LCEL** (LangChain Expression Language), que é um recurso de linguagem declarativa do LangChain para criar cadeias de execução. 

Uma cadeia (_chain_) no LangChain é uma estrutura de orquestração que define um fluxo de execução composto por múltiplos componentes. É um objeto que implementa uma sequência lógica onde o resultado de um estágio serve, automaticamente, como entrada para o próximo. Aqui vamos orquestrar, de forma sequencial e organizada, o fluxo que vai desde o _prompt_, passando pelo modelo de linguagem (LLM), até o tratamento da saída, tornando a aplicação mais robusta e profissional. 

### _Output Parsers_

#### Resposta como _string_

Então começamos pelo "tratamento da saída", vamos garantir que a resposta retornada esteja no formato que queremos. Para isso vamos usar o mais simples dos [_Output Parsers_](https://medium.com/data-and-beyond/output-parsers-in-langchain-b2e0db20880f), a classe [`StrOutputParser`](https://reference.langchain.com/python/langchain-core/output_parsers/string/StrOutputParser), que extrai o conteúdo de texto da resposta do LLM:

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser # permite controlar o formato da resposta
from dotenv import load_dotenv
import os

load_dotenv()
api_key       = os.getenv("OPENAI_API_KEY")
model_version = "gpt-6-luna"
```

Também vamos modificar o modelo de _prompt_, não apenas mudando seu conteúdo, mas seu funcionamento. Não vou mais passar as variáveis para ele, mas apenas deixar indicado quais são as variáveis que espero:

```python
# inicializa o prompt template definindo as variaveis necessarias
prompt = PromptTemplate(
    template="Sugira uma cidade para minha viagem, dado meu interesse por {interesse}.",
    input_variables=["interesse"]
)
prompt
```

Feito isso, podemos definir a cadeia e chamar `invoke` a partir dela, e o que passamos é um dicionário em que cada chave é uma variável de _input_ para _prompt_:

```python
# inicializa objeto para requisicao ao modelo de linguagem
llm = ChatOpenAI(model=model_version, api_key=api_key, cache=False)

# instancia parser para string
parser = StrOutputParser()

# define cadeia
chain = prompt | llm | parser

chain
```

```python
# dispara execucao da cadeia
city_response = chain.invoke({"interesse": "praias"})
type(city_response)
```

```python
# note que a resposta ja eh uma string, pois a saida foi transformada no final da cadeia
print(city_response)
```

A cadeia acima opera da seguinte forma:

- **Transformação do _input_:** o componente de _Prompt_ recebe um dicionário com variáveis e transforma esses dados em um objeto estruturado que o LLM consegue processar.
- **Processamento do Modelo:** o LLM recebe esse objeto estruturado, processa as informações e gera uma resposta, que é retornada como um objeto de mensagem contendo diversos metadados, e não apenas o texto puro.
- **Formatação da Saída (_Output Parser_):** o parser atua como um transformador final. Ele recebe do LLM um objeto de resposta complexo, e então extrai, filtra ou converte esse resposta para um formato específico, nesse caso, uma _string_ simples (usando `StrOutputParser`).

O operador `|` (_pipe_) permite a composição linear e modular de diferentes etapas de processamento, estabelecendo essa conexão formal entre eles. Ele cria um fluxo onde, ao chamar o método `invoke` na cadeia, o LangChain gerencia internamente a passagem de dados através desses componentes, de forma que a saída de uma função seja automaticamente a entrada da seguinte, sem que seja necessário escrever manualmente o código para tratar a transição entre cada etapa.

Em vez de invocar o _prompt_, capturar o resultado, passar para o modelo, capturar a resposta, acessar o atributo `content` e então formatar, apenas definimos a cadeia uma única vez, e ela encapsula todo esse processo em uma única interface, tornando o código mais modular, legível e fácil de manter. Essa abordagem encoraja a separação clara de responsabilidades, facilitando futuras manutenções e adaptações.

#### Resposta como JSON

Em algumas situações, uma _string_ pode não ser o formato mais adequado para uma resposta, nesses casos precisamos definir um _parser_ que tranforme a resposta para o esse formato mais adequado. Um formato bastante comum é o JSON, que pode ser obtido através do [`JsonOutputParser`](https://reference.langchain.com/python/langchain-core/output_parsers/json/JsonOutputParser).

Além disso, podemos combiná-lo ser com o [Pydantic](https://pydantic.dev/docs/validation/latest/get-started/), uma biblioteca utilizada para validação de dados e gestão de configurações. O [Pydantic](https://medium.com/@habbema/pydantic-23fe91b5749b) permite definir a estrutura de modelos através de classes, garantindo que os dados sigam os tipos e restrições especificados, facilitando a manipulação e a serialização de objetos, como no caso do controle de formatos de saída JSON. E modelo nesse contexto se refere ao "modelo do formato" de resposta esperada.

Para definir a estrutura da reposta com Pydantic, vamos utilizar algumas de suas classes:

- `BaseModel`: define a estrutura do modelo de formato que queremos receber
- `Field`: permite definir os campos do modelo de resposta de acordo com o tipo desses dados que queremos receber.

Com isso, podemos criar uma classe descrevendo o formato de dados que queremos receber. Essa classe vai extender a `BaseModel`

```python
from langchain_core.output_parsers import JsonOutputParser
from pydantic import BaseModel, Field

# define classe que representa formato esperado para resposta
class Destination(BaseModel):
    city:   str = Field("A cidade recomendada para visitar")
    reason: str = Field("motivo que pelo qual é interessante visitar essa cidade")

# define parser tranformar a reposta de acordo com o formato esperado
parser_destination = JsonOutputParser(pydantic_object=Destination)

# atualiza prompt template com o formato de saida esperado
prompt_destination = PromptTemplate(
    template="""
        Sugira uma cidade para minha viagem, dado meu interesse por {interesse}.
        {formato_saida}
    """,
    input_variables=["interesse"],
    partial_variables={"formato_saida": parser_destination.get_format_instructions()}
)
```

```python
print(prompt_destination.get_format_instructions())
```

Note que precisamos informar ao _prompt template_ como queremos receber a reposta, e fazemos isso através das "variáveis parciais". Essas variáveis são passadas através de um dicionário.

Após isso, redefinimos e disparamos a execução da cadeia:

```python
# define cadeia
chain = prompt_destination | llm | parser_destination

# dispara execucao da cadeia
city_response_json = chain.invoke({"interesse": "praias"})
type(city_response_json)
```

```python
print(city_response_json)
```

Podemos importar o objeto [`set_debug`](https://reference.langchain.com/python/langchain-core/globals/set_debug) do pacote de configurações globais e ativar o "debugger". Para visualizarmos como a cadeia é executada:

Veja que, no fim das contas, a _string_ do "_prompt_ do formato" foi concatenada a _string_ do _prompt_ da pergunta do usuário e tudo foi enviado para o LLM na mesma requisição. Mas então porque não passamos o formato de saída em `input_variables` de uma vez? A diferença fundamental reside no momento e na forma como essas variáveis são processadas antes de serem enviadas ao LLM. As `input_variables` são parâmetros dinâmicos que o usuário fornece no momento em que invoca a cadeia, ou seja, são os dados que mudam conforme a interação. Já as `partial_variables` são definidas antecipadamente, as utilizamos quando queremos "injetar" um valor fixo ou o resultado de uma função no _template_ (como o `get_format_instructions`, por exemplo), sem a necessidade de o usuário ou a execução da cadeia precisar fornecer essa informação como um argumento de entrada a cada nova chamada.

Apesar do resultado final no _prompt_ enviado ao LLM ser, de fato, uma _string_ concatenada, o uso de `partial_variables` permite organizar melhor as responsabilidades no código: a instrução de formatação é mantida separada da lógica de entrada do usuário. Isso facilita a manutenção e garante que as instruções técnicas necessárias para o modelo entender o formato JSON estejam sempre presentes no _template_ de forma automatizada.

### Sequencia de cadeias

Agora que implementamos uma cadeia do início ao fim, vamos explorar uma "sequencia de cadeias", em que concatenamos mais de uma cadeia. Vamos concatenar cadeias que nos tragam mais informações úteis para montar um roteiro de viagens, como por exemplo, descobrir bons restaurantes e passeios culturais na cidade sugerida pela primeira cadeia.

Vamos concatenar duas novas cadeias a "cadeia da cidade", sendo a do meio para nos indicar restaurantes e a última para nos sugerir passeios culturais. Começamos criando uma classe em que especificamos o formato da resposta da "cadeia de resutaurantes", que será bastante parecido com a primeira classe, definimos o _parser_ que recebe essa nova classe, e então, passamos para o _prompt_ dessa nova pergunta:

```python
# define classe que representa formato esperado para resposta dos restaurantes
class Restaurants(BaseModel):
    city:        str = Field("A cidade recomendada para visitar")    # essa informacao sera passada para a proxima cadeia
    restaurants: str = Field("Restaurantes recomendandos na cidade")
    
# define parser para a reposta dos restaurantes
parser_restaurants = JsonOutputParser(pydantic_object=Restaurants)

# define prompt template para a pergunta dos restaurantes
prompt_restaurants = PromptTemplate(
    template="""
        Sugira restaurantes populares entre locais em {city}.
        {formato_saida}
    """,
    partial_variables={"formato_saida": parser_restaurants.get_format_instructions()}
)
```

Note que o parâmetro `input_variables` foi removido do _template_ `prompt_restaurants`, pois como as cadeias estarão interligadas, o objeto resultante da execução de uma cadeia é passado como entrada para a próxima. Então como a primeira cadeia retorna um objeto que contém uma chave chamada `city`, e o _prompt_ da segunda cadeia contém a variável `{city}`, o LangChain é capaz de realizar essa substituição.

E a última cadeia não terá um formato customizado, vamos definir que sua resposta seja apenas uma _string_:

```python
# define template para pergunta de passeios culturais
prompt_culture = PromptTemplate(
    template="Sugira atividades e locais culturais em {city}"
)

# define parser para a reposta de passeios culturais
parser_culture = StrOutputParser()
```

```python
# define cada uma das cadeias
chain_1 = prompt_destination | llm | parser_destination
chain_2 = prompt_restaurants | llm | parser_restaurants
chain_3 = prompt_culture     | llm | parser_culture

# define sequenia de cadeias
chain   = (chain_1 | chain_2 | chain_3)

# dispara execucao da sequencia de cadeias
response = chain.invoke({"interesse": "praias"})
type(response)
```

```python
print(response)
```

#### Rótulos "casados" entre cadeias

Como já mencionado, essa passagem de parâmetros entre as cadeias funciona porque são utilizados os mesmos rótulos na resposta de uma cadeia e no _input_ da seguinte. Se isso não acontecer, se a primeira cadeia retorna um campo chamado `colors` e a segunda espera um parâmetro chamado `cores`, o fluxo não funcionará automaticamente, o motor do LangChain não sabe que o conteúdo de `colors` deve ser mapeado para a variável `cores` e receberemos um erro do tipo:

> `KeyError: "Input to PromptTemplate is missing variables {'cores'}.  Expected: ['cores'] Received: ['colors']`

Para resolver isso, seria necessário criar uma etapa intermediária que utilize uma função ou um componente como o `RunnablePassthrough`, que permita manipular a saída da primeira cadeia antes de entregá-la à segunda. O dicionário de entrada precisaria ter sua a chave `colors` "renomeada" para `cores`, garantindo que a segunda cadeia receba exatamente o que ela espera.

Abaixo foi implementado um exemplo disso.

```python
# define formato esperado da resposta e parser
class Colors(BaseModel):
    colors: str = Field("As cores sugeridas")

parser = JsonOutputParser(pydantic_object=Colors)

# define templates dos prompts
prompt_begin  = PromptTemplate(template="Sugira uma nova cor e concatene com {initial_color}. {saida}", input_variables=["initial_color"], partial_variables={"saida": parser.get_format_instructions()})
prompt_middle = PromptTemplate(template="Sugira uma nova cor e concatene com {colors}. {saida}", partial_variables={"saida": parser.get_format_instructions()})
prompt_end    = PromptTemplate(template="Sugira uma nova cor e concatene com {colors}.")

# define as cadeias
colors_begin  = prompt_begin  | llm | parser
colors_middle = prompt_middle | llm | parser
colors_end    = prompt_end    | llm | StrOutputParser()

# concate as cadeias
response_colors = (colors_begin | colors_middle | colors_end).invoke({"initial_color": "Azul"})
print(response_colors)
```

```python
from langchain_core.runnables import RunnablePassthrough

# redefine template, agora espera input com rotulo `cores` e nao `colors`
prompt_end = PromptTemplate(template="Sugira uma nova cor e concatene com {cores}.")
colors_end = prompt_end | llm | StrOutputParser() # sera que precisa disso?

translate_response = RunnablePassthrough.assign(
    cores=lambda x: x["colors"]
)

response_colors = (colors_begin | colors_middle | translate_response | colors_end).invoke({"initial_color": "Azul"})
print(response_colors)
```
