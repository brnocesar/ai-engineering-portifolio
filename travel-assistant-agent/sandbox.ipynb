## Conectando direto na API da OpenAI

```python
from openai import OpenAI
from dotenv import load_dotenv
import os
```

```python
# carrega chave da api para variaveis de ambiente do sistema
load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")

model_version = "gpt-6-luna"
```

```python
# monta prompt para interacao com API da OpenAI
numero_dias     = 7
numero_criancas = 2
atividade       = "música"
prompt          = f"Crie um roteiro de viagem de {numero_dias} dias, para uma familia com {numero_criancas} crianças, que gosta de {atividade}"
```

```python
# define client que vai permitir interacao com API da OpenAI
client = OpenAI(api_key=api_key)
client
```

Agora que temos o _client_ definido, vamos testar algumar requisições para a [API](https://developers.openai.com/api/docs/guides/migrate-to-responses), primeiro com _chat completion_ e depois uma "resposta".

### _Chat Completion_

Esse formato é uma estrutura de interação muito específica da API da OpenAI. O _chat completion_ é o recurso projetado para receber mensagens de um histórico ou uma pergunta e retornar uma resposta gerada pelo modelo. É uma forma de comunicação que segue a lógica interna e as nomenclaturas da própria OpenAI.

```python
# cria chat completion para interacao
# parametro messages eh uma lista que precisa de ao menos um item, mas geralmente passamos dois
response = client.chat.completions.create(
    model=model_version,
    messages=[
        {"role": "system", "content": "Você é um assistente de roteiro de viagens."},   # um que indica como o sistema deve se comportar
        {"role": "user",   "content": prompt}                                           # outro com a pergunta do usuario
    ],
    store=False
)
type(response)
```

```python
for r in response:
    print(r)
```

```python
type(response.choices), len(response.choices)
```

```python
for choice in response.choices:
    for property in choice:
        print(property)
```

```python
type(response.choices[0].message)
```

```python
print(response.choices[0].message.content.replace("; ", ";\n"))
```

### Resposta

```python
response = client.responses.create(
    model=model_version,
    input=prompt, # recebe apenas o prompt
    store=False
)

type(response)
```

```python
for r in response:
    print(r)
```
